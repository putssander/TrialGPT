```python
import json
import pandas as pd

# File paths
dataset="mstro"
queries_file = f"../dataset/{dataset}/queries.jsonl"
corpus_file = f"../dataset/{dataset}/corpus.json"
output_tsv = f"../dataset/{dataset}/qrels/test.tsv"


# Load queries.jsonl
queries = []
with open(queries_file, "r", encoding="utf-8") as f:
    for line in f:
        data = json.loads(line.strip())  # Load each line as a JSON object
        queries.append(data["_id"])  # Extract query ID

# Load corpus.json
with open(corpus_file, "r", encoding="utf-8") as f:
    corpus = json.load(f)  # Load JSON list

# Extract corpus IDs
corpus_ids = [doc["_id"] for doc in corpus]

# Create DataFrame
data = []
for query_id in queries:
    for corpus_id in corpus_ids:
        data.append([query_id, corpus_id, 1])  # Assign score of 1

df = pd.DataFrame(data, columns=["query-id", "corpus-id", "score"])

# Save to TSV
df.to_csv(output_tsv, sep="\t", index=False)

print(f"TSV file saved as {output_tsv}")
```
