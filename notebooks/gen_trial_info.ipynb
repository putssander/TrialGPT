```python
import json

dataset = "mstro"
path_corpus = f"../dataset/{dataset}/corpus.jsonl"
path_trial_info = f"../dataset/trial_info.json"

# Dictionary to store trial info
trial_info = {}

# Process each line in the corpus.jsonl file
with open(path_corpus, 'r') as file:
    for line in file:
        # Parse each line as a JSON object
        item = json.loads(line)
        trial_id = item['_id']
        
        # Create entry with the required fields
        trial_entry = {
            'NCTID': trial_id,
            'brief_title': item['metadata'].get('brief_title', ''),
            'diseases_list': item['metadata'].get('diseases_list', []),
            'brief_summary': item['metadata'].get('brief_summary', ''),
            # Add these additional fields that might be useful
            'title': item.get('title', ''),
            'phase': item['metadata'].get('phase', ''),
            'drugs_list': item['metadata'].get('drugs_list', []),
            'diseases': item['metadata'].get('diseases', ''),
            'enrollment': item['metadata'].get('enrollment', ''),
            'inclusion_criteria': item['metadata'].get('inclusion_criteria', ''),
            'exclusion_criteria': item['metadata'].get('exclusion_criteria', '')
        }
        
        # Add to the trial_info dictionary
        trial_info[trial_id] = trial_entry

# Save to trial_info.json
with open(path_trial_info, 'w', encoding='utf-8') as file:
    json.dump(trial_info, file, ensure_ascii=False, indent=4)

print(f"Converted {len(trial_info)} trials from corpus.jsonl to trial_info.json")
```
