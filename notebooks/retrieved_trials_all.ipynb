```python
path_result_retrieval_keywords = "../results/retrieval_keywords_ollama:phi4_mstr.json"
results_name = "phi4-turbo" ## turbo for code compatibility
dataset = "mstro"
path_id2queries = f"../dataset/{dataset}/id2queries.json"
path_trials = f"../dataset/{dataset}/corpus.xlsx"


import json
import pandas as pd

trials_df = pd.read_excel(path_trials)
# Convert the DataFrame to a JSON list
trials_json = trials_df.to_json(orient='records', force_ascii=False)
# Display or save the JSON output
print(trials_json)
```

```python
# Open and load the JSON file
with open(path_id2queries, 'r') as file:
    data = json.load(file)

# Print the loaded data
retrieved_trials = []
for item in data:
    entry = data[item]
    
    new_item = {}
    new_item["0"] = json.loads(trials_json)
    new_item["patient_id"] = item
    new_item["patient"] = entry["raw"]
 
    retrieved_trials.append(new_item)

print(retrieved_trials)
path_trials = f"../dataset/{dataset}/retrieved_trials.json"
with open(path_trials, 'w', encoding='utf-8') as file:
    json.dump(retrieved_trials, file, ensure_ascii=False, indent=4)
```
